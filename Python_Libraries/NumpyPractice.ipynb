```python
# Create a 4x5 array of random integers between 1 and 100 and store it as arr. Use it in the next sections.
arr4 = np.random.randint(1,100,(4,5))
arr4
```

```python
# Print the shape, dtype, number of dimensions, and size of arr4.
print(arr4.shape) #shape - (row,col) of the array
print(arr4.dtype) #dtype - data types of array elements
print(arr4.ndim) #ndim - dimension of array 1d , 2d, 3d ....nd
print(arr4.size) #size - no of elements
```

# **Section B : Indexing and slicing**

```python
#Get the first row of arr4, the last column, and the element at row 2, column 3.
print(arr4[0,:])
print(arr4[:,4])
print(arr4[1,2])
```

```python
# Get the top-left 2x2 block.
print(arr4[:2,:2])
```

```python
#Get every second column.
print(arr4[:,1])
```

```python
# Reverse the order of the rows.
print(arr4[::-1,:])
```

```python
# Get all elements greater than 50 as a 1D array (boolean masking).
print(arr4[arr4>50])
```

```python
# Replace every value below 20 with 0, working on a copy so arr stays intact. Check why .copy() matters by trying it without one.
arr4_copy=arr4.copy()
# here in numpy .copy() create an independent copy of the array
arr4_copy[arr4_copy<20]=0
arr4_copy
```

# **Section C: Math and aggregation**

```python
# Find the mean, median, min, max, and standard deviation of arr.
print("Mean : ",np.mean(arr4))
print("Median : ",np.median(arr4))
print("Minimum value : ",np.min(arr4))
print("Maximum value : ",np.max(arr4))
print("Standard Deviation : ",np.std(arr4))
```

```python
# Find the column-wise sum and the row-wise mean (use the axis argument).
print("Column wise sum : ",np.sum(arr4,axis=0))
print("Row wise mean : ",np.mean(arr4,axis=1))
# • axis=0 (Vertical / Downwards)
# • axis=1 (Horizontal / Across)
```

```python
# Find the index of the maximum value in each row (argmax).
print("Row wise max value index : ",np.argmax(arr4,axis=1))
```

```python
# Subtract each column’s mean from that column (centering the data), using broadcasting with no loops.
col_means=np.mean(arr4,axis=0)
arr4_centered=arr4-col_means
arr4_centered
```

```python
# Create two arrays a = np.array([1, 2, 3]) and b = np.array([4, 5, 6]). Compute element-wise product, dot product, and a squared.
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
print(a*b)
print(np.dot(a,b))
print(a**2)
```

#**Section D: Reshaping and combining**

```python
# Create an array of numbers 1 to 12 and reshape it into 3x4, then 2x6, then 4x3.
arr5=np.arange(1,13)
r1= arr5.reshape(3,4)
print(r1)
r2=arr5.reshape(2,6)
print(r2)
r3=arr5.reshape(4,3)
print(r3)
```

```python
# Flatten a 2D array back to 1D. (r1 -> 1d)
r1_flat=r1.flatten()
r1_flat
```

```python
# Stack two arrays vertically and horizontally (vstack, hstack).
a1=np.arange(1,13).reshape(3,4)
a2=np.arange(13,21).reshape(2,4)
print(np.vstack((a1,a2)))
```

```python
a3=np.arange(1,13).reshape(3,4)
a4=np.arange(13,19).reshape(3,2)
print(np.hstack((a3,a4)))
```
