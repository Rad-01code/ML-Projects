# **Section B: Inspecting the data**


```python
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/tips.csv"
df = pd.read_csv(url)
```

```python
# Show the first 5 and last 3 rows.
print("First 5 rows : ")
print(df.head())
print("---------------------------------------------------------------")
print("Last 3 rows : ")
print(df.tail(3))
```

```python
# Print the shape, column names, and data types.
print("Shape of dataset : ",df.shape)
print("Column names : ",df.columns)
print("Data Types : \n",df.dtypes)
```

```python
# Use df.info() and df.describe(). Which column has the highest mean, and which has the highest max?
df.info()
```

```python
df.describe()
```

total bill has the highest mean and highest max

```python
# How many unique values does day have, and how many rows for each (value_counts)?
print(df['day'].unique())
print(df['day'].value_counts())
```

### Note:
*   describe() only covers numeric columns. To include text columns too, use df.describe(include='all'), which adds count, unique, top, and freq for the categorical ones.
*   nunique() = count of distinct values; unique() = the values themselves.



# **Section C: Selecting columns and rows**

```python
# Select just the tip column, first as a Series (df["tip"]) and then as a DataFrame (df[["tip"]]). Print the type of each.
print(type(df['tip']))
print(type(df[['tip']]))
```

```python
# Select the columns total_bill, tip, and size together.
print(df.loc[:,['tip','total_bill','size']])
```

```python
# Use iloc to get the first row, rows 5 to 9, and the element at row 3, column 2.
print("First row : \n",df.iloc[0])
print("-------------------------------------------------")
print("Rows  5 to 9 : \n",df.iloc[5:10])
print("-------------------------------------------------")
print("Row 3 col 2 element : ",df.iloc[3,2])
```

loc includes the end label; iloc excludes the end position

```python
# Use loc to get rows 0 to 4 for only the columns total_bill and tip. Notice whether the end of the range is included, then compare with iloc[0:4].
print(df.loc[0:4,['total_bill','tip']])
```

```python
# Get every 10th row of the DataFrame using slicing.
print(df.iloc[::10])
```

# **Section D: Filtering rows**

```python
# Get all bills where total_bill is greater than 40.
df[df['total_bill']>40]
```

```python
#Get all Sunday bills (day == "Sun") for dinner (time == "Dinner") using &.
df[(df['day']=='Sun')& (df['time']=='Dinner')]
```

```python
# Get rows where day is Sat or Sun, first using |, then using .isin(["Sat", "Sun"]).
df[(df['day']=='Sun') | (df['day']=='Sat')]
```

```python
df[df['day'].isin(['Sat','Sun'])]
```

```python
# Get bills for smokers where the tip was above 5, and show only total_bill, tip, and day (combine a filter with loc).
ans = df[(df['smoker']=='Yes') &  (df['tip']>5)].loc[:,['total_bill','tip','day']]
print(ans)
```

```python
# How many rows are in your answer to question 16? Which method gives you that number (len or shape)?
print(len(ans))
```

```python
# New column: create tip_pct = tip / total_bill * 100. Find the row with the highest tip percentage (idxmax).
df['tip_pct']=df['tip']/df['total_bill'] * 100
df.head()
```

```python
df.loc[df['tip_pct'].idxmax()]
```

```python
# Tidy view: show the 10 largest bills with only the columns total_bill, tip, tip_pct, and day (try sort_values).
sorted_bill = df.sort_values(by='total_bill',ascending=False)
sorted_bill.head(10)
```
